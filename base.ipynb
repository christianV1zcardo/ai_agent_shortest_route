```python
from pathlib import Path
import osmnx as ox
import networkx as nx
import sklearn

ox.settings.use_cache = False
```

```python
place = "San Isidro, Lima, Peru"

G = ox.graph.graph_from_place(place, network_type='drive')

ox.plot.plot_graph(G)
```

```python
list(G.nodes)[5:10]
type(G)
```

```python
# Latitud y longitud convertida a nodos más cercanos
origen_lat, origen_lon = -12.10015, -77.05240
destino_lat, destino_lon = -12.10277, -77.03774

origen_node = ox.distance.nearest_nodes(G, origen_lon, origen_lat)
destino_node = ox.distance.nearest_nodes(G, destino_lon, destino_lat)

print(origen_node)
print(destino_node)
```

```python
# Longitud de la ruta
ruta_longitud = nx.shortest_path_length(G, origen_node, destino_node, weight='length')

# Ruta nodo a nodo
ruta = nx.shortest_path(G, origen_node, destino_node, weight='length')

print(ruta_longitud)
ruta
```
